## 🧭 Módulo 0: ¿Qué es Diseño y Arquitectura? Los Dos Valores del Software
*Referencia: Capítulos 1 y 2 de "Arquitectura Limpia"*

### 1. ¿Hay diferencia entre Diseño y Arquitectura?
Existe una confusión común en la industria: creer que la *arquitectura* es algo de alto nivel reservado para diagramas UML, y que el *diseño* es el detalle de bajo nivel del programador.

**Uncle Bob destruye esa separación:**
> *"No hay diferencia. Ninguna en absoluto. La palabra 'diseño' a menudo se usa en el contexto de decisiones de bajo nivel, y 'arquitectura' en el de alto nivel. Pero cuando las observas de cerca, forman una trama continua de decisiones sin una línea divisoria clara."*

### 2. El Objetivo Supremo
> **"La meta de la arquitectura de software es minimizar los recursos humanos necesarios para construir y mantener el sistema requerido."**
El síntoma de una arquitectura degradada es la curva de esfuerzo: cada nueva funcionalidad cuesta el doble de tiempo y requiere el doble de desarrolladores que la anterior.

### 3. El Cuento de los Dos Valores del Software
1. **El Valor de Comportamiento (Funcionalidad)**: La máquina hace lo que el usuario necesita hoy. Genera ingresos en el presente.
2. **El Valor de Estructura (Arquitectura)**: El software debe ser **"SOFT"** (suave, flexible, maleable). Debe ser fácil de cambiar ante nuevas necesidades del mercado mañana.

#### La Matriz de Eisenhower (Urgente vs. Importante):
* **Cuadrante 1: Urgente e Importante** (Incendios críticos).
* **Cuadrante 2: No urgente pero Importante** (¡Aquí vive la Arquitectura!).
* **Cuadrante 3: Urgente pero No importante** (La feature rápida que el cliente pide "para ayer").
* **Cuadrante 4: Ni urgente ni importante**.

> ⚠️ **El gran error de los equipos:** Confundir lo Urgente con lo Importante.
> Cuando el negocio pide una funcionalidad con prisa (Cuadrante 3), los desarrolladores a menudo sacrifican la arquitectura (Cuadrante 2). Al hacer esto repetidamente, el software se vuelve rígido y frágil.

### 💻 Código Demostrativo: La Pesadilla del Acoplamiento
Observa cómo luce un código espagueti donde la entrada/salida, las reglas de negocio, la persistencia y la presentación están mezcladas:


````python
# EJEMPLO DE CÓDIGO ACOPLADO (Sin arquitectura)
base_de_datos_ficticia = {}

def procesar_alta_empleado_espagueti(nombre: str, salario_base: float, es_gerente: bool):
    # 1. Validación de entrada (I/O)
    if not nombre or salario_base <= 0:
        print("[ERROR DE VISTA] Datos inválidos.")
        return False
    
    # 2. Regla de negocio crítica
    salario_final = salario_base
    if es_gerente:
        salario_final += salario_base * 0.20  # Bono gerente 20%
    if salario_final > 5000:
        salario_final -= salario_final * 0.10 # Retención impositiva 10%
        
    # 3. Persistencia directa
    nuevo_id = len(base_de_datos_ficticia) + 1
    base_de_datos_ficticia[nuevo_id] = {"nombre": nombre, "salario": salario_final}
    
    # 4. Presentación
    print(f"=== EMPLEADO CREADO === ID: #{nuevo_id:04d} | Salario Neto: ${salario_final:,.2f}")
    return True

procesar_alta_empleado_espagueti("Carlos Gómez", 6000.0, es_gerente=True)
````

## 🧱 Módulo 1: Los Principios SOLID Vistos con Ojos de Arquitecto
*Referencia: Parte III (Capítulos 7 al 11) de "Arquitectura Limpia"*

Los principios **SOLID** suelen enseñarse para organizar clases pequeñas, pero Uncle Bob enfatiza que **su verdadero propósito es arquitectónico**: establecen cómo deben interactuar los límites de componentes y cómo aislar los cambios.

### 1. SRP: Principio de Responsabilidad Única
> *"Un módulo debe ser responsable ante uno, y solo uno, **actor**."*
Un **actor** es un grupo de usuarios o stakeholders (CFO, COO, CTO) que solicitan cambios por sus propias razones. Si Finanzas y Operaciones comparten la misma clase, un cambio de Finanzas puede romper los reportes de Operaciones.

### 2. OCP: Principio Abierto/Cerrado
> *"Un artefacto de software debe estar abierto para su extensión, pero cerrado para su modificación."*
Si agregar una nueva pasarela de pago te obliga a modificar 15 archivos existentes con `if-elif-else`, la arquitectura falló. Se deben agregar nuevas características escribiendo **nuevo código**, no modificando código probado.

### 3. LSP: Principio de Sustitución de Liskov
> *"Los subtipos deben ser sustituibles por sus tipos base sin alterar la corrección del programa."*
El contrato definido por una abstracción debe ser respetado estrictamente por todas las implementaciones concretas.

### 4. ISP: Principio de Segregación de Interfaces
> *"Ningún cliente debe ser forzado a depender de métodos que no utiliza."*
Evita interfaces infladas. Si un cliente solo necesita cobrar, no lo obligues a implementar métodos de auditoría o reportes contables.

### 5. DIP: Principio de Inversión de Dependencias
> 1. *"Los módulos de alto nivel no deben depender de los de bajo nivel. Ambos deben depender de abstracciones."*
> 2. *"Las abstracciones no deben depender de los detalles. Los detalles deben depender de las abstracciones."*
El negocio no debe llamar a MySQL ni a Stripe directamente. Con DIP, la base de datos se convierte en un plugin que implementa un contrato dictado por el negocio.

### 💻 Código Demostrativo: Violación de DIP e ISP vs. Inversión Limpia
Observemos cómo DIP invierte el control mediante una interfaz (`Protocol` o `ABC` en Python):


````python
# ANTES: Dependencia directa (Bajo nivel controla a Alto nivel)
class TwilioSMSClient:
    def send_sms_via_http(self, phone: str, text: str):
        print(f"[Twilio Red] Enviando SMS a {phone}: {text}")

class GestorPedidosMal:
    def __init__(self):
        self.sms_client = TwilioSMSClient() # <- ¡Acoplamiento rígido!

    def completar_pedido(self, cliente_tel: str, total: float):
        mensaje = f"Su pedido de ${total:.2f} ha sido confirmado."
        self.sms_client.send_sms_via_http(cliente_tel, mensaje)

# DESPUÉS: Con DIP (Inversión de Dependencias)
class Notificador(ABC):
    @abstractmethod
    def notificar(self, destinatario: str, mensaje: str) -> None:
        pass

class GestorPedidosLimpio:
    def __init__(self, notificador: Notificador):
        self.notificador = notificador # Inyección de dependencias

    def completar_pedido(self, cliente_tel: str, total: float):
        mensaje = f"Su pedido de ${total:.2f} ha sido confirmado."
        self.notificador.notificar(cliente_tel, mensaje)

class TwilioAdapter(Notificador):
    def notificar(self, destinatario: str, mensaje: str) -> None:
        print(f"[Twilio Cloud] SMS a {destinatario}: {mensaje}")

class EmailAdapter(Notificador):
    def notificar(self, destinatario: str, mensaje: str) -> None:
        print(f"[SendGrid Email] Enviando correo a {destinatario}: {mensaje}")

gestor_sms = GestorPedidosLimpio(TwilioAdapter())
gestor_sms.completar_pedido("+541199999", 1500.0)

gestor_email = GestorPedidosLimpio(EmailAdapter())
gestor_email.completar_pedido("cliente@email.com", 1500.0)
````

* Son estructuras de datos planas (sin métodos ni lógica).
* **Regla de oro**: Nunca expongas tus entidades directamente a la UI o API web, ni pases el request HTTP crudo al caso de uso.

````python
# 1. ENTIDAD (Pura, sin dependencias)
@dataclass
class CuentaBancaria:
    numero_cuenta: str
    titular: str
    saldo: Decimal = Decimal("0.00")
    limite_descubierto: Decimal = Decimal("500.00")

    def depositar(self, monto: Decimal) -> None:
        if monto <= Decimal("0.00"):
            raise ValueError("El monto a depositar debe ser mayor a cero.")
        self.saldo += monto

    def extraer(self, monto: Decimal) -> None:
        if monto <= Decimal("0.00"):
            raise ValueError("El monto a extraer debe ser mayor a cero.")
        if (self.saldo - monto) < -self.limite_descubierto:
            raise ValueError("Fondos insuficientes: excede límite de descubierto.")
        self.saldo -= monto

# 2. DTOs planos
@dataclass(frozen=True)
class TransferirRequest:
    cuenta_origen_id: str
    cuenta_destino_id: str
    monto: Decimal

@dataclass(frozen=True)
class TransferirResponse:
    exito: bool
    mensaje: str
    nuevo_saldo_origen: Optional[Decimal] = None

cuenta = CuentaBancaria("001-A", "Ana Rossi", Decimal("100.00"))
cuenta.extraer(Decimal("300.00"))
print(f"Saldo con descubierto: ${cuenta.saldo}")
````

> **"Las dependencias en el código fuente SOLO pueden apuntar hacia adentro, hacia políticas de más alto nivel.**

Si el Caso de Uso necesita guardar datos en PostgreSQL:
1. El Caso de Uso define una interfaz abstracta (**Puerto de Salida**) en su propio círculo:
   `class CuentaRepositoryPort(ABC): ...`
2. En la capa externa, la base de datos implementa esa interfaz:
   `class PostgresCuentaRepository(CuentaRepositoryPort): ...`
3. En ejecución, el flujo de control va hacia afuera, pero la dependencia del código fuente apunta hacia adentro.

### 💻 Código Demostrativo: Definiendo el Puerto y el Caso de Uso


````python
# 1. PUERTO DE SALIDA (Vive dentro de la capa del caso de uso)
class CuentaRepositoryPort(ABC):
    @abstractmethod
    def obtener_por_numero(self, numero_cuenta: str) -> Optional[CuentaBancaria]:
        pass

    @abstractmethod
    def guardar(self, cuenta: CuentaBancaria) -> None:
        pass

# 2. CASO DE USO (Orquestador puro)
class TransferirDineroUseCase:
    def __init__(self, repositorio: CuentaRepositoryPort):
        self.repositorio = repositorio

    def ejecutar(self, request: TransferirRequest) -> TransferirResponse:
        if request.monto <= Decimal("0.00"):
            return TransferirResponse(exito=False, mensaje="El monto debe ser positivo.")
        
        origen = self.repositorio.obtener_por_numero(request.cuenta_origen_id)
        if not origen:
            return TransferirResponse(exito=False, mensaje="Cuenta origen no encontrada.")

        destino = self.repositorio.obtener_por_numero(request.cuenta_destino_id)
        if not destino:
            return TransferirResponse(exito=False, mensaje="Cuenta destino no encontrada.")

        try:
            origen.extraer(request.monto)
            destino.depositar(request.monto)
        except ValueError as err:
            return TransferirResponse(exito=False, mensaje=str(err))

        self.repositorio.guardar(origen)
        self.repositorio.guardar(destino)

        return TransferirResponse(
            exito=True, 
            mensaje="Transferencia completada exitosamente.",
            nuevo_saldo_origen=origen.saldo
        )

print("Caso de uso compilado. No importa si la BD es SQL, Mongo o memoria.")
````

## 🎭 Módulo 4: Adaptadores de Interfaz y el Patrón "Humble Object" (Objeto Humilde)
*Referencia: Capítulos 23 y 24 de "Arquitectura Limpia"*

### 1. El Patrón Humble Object (Objeto Humilde)
> *"Cuando una parte del software es difícil de probar (GUI, consola, driver SQL), **divídela en dos objetos**:"*
> 1. **El Objeto Inteligente**: Contiene toda la lógica de formateo, colores y condiciones. 100% testeable sin entorno gráfico.
> 2. **El Objeto Humilde**: Es tan simple y tonto que no tiene sentido probarlo. Solo toma datos ya masticados y los pasa a la pantalla o socket.

```
 ┌────────────────────────┐      ┌─────────────────────────┐      ┌─────────────────────────┐
 │      CASO DE USO       │ ───► │  PRESENTER (Inteligente)│ ───► │   VISTA (Objeto Humilde)│
 │ (Devuelve ResponseDTO) │      │ (Formatea a ViewModel)  │      │ (Solo imprime / render) │
 └────────────────────────┘      └─────────────────────────┘      └─────────────────────────┘
```

### 2. Presenters y ViewModels
* El Caso de Uso devuelve datos crudos: `total = Decimal("14400.00")`, `aprobado = True`.
* El **Presenter** toma ese DTO y lo transforma en un **ViewModel**:
  - `total_formateado = "$14.400,00 ARS"`, `estado_texto = "APROBADO"`, `color_alerta = "VERDE"`.
* La **Vista** es humilde: **no contiene condicionales `if` de negocio**. Solo dibuja lo que dice el ViewModel.

````python
import json

# ADAPTADOR 1: Memoria
class EnMemoriaCuentaRepository(CuentaRepositoryPort):
    def __init__(self):
        self._tabla: Dict[str, CuentaBancaria] = {}

    def obtener_por_numero(self, numero: str) -> Optional[CuentaBancaria]:
        return self._tabla.get(numero)

    def guardar(self, cuenta: CuentaBancaria) -> None:
        self._tabla[cuenta.numero_cuenta] = cuenta

# ADAPTADOR 2: Archivo JSON
class JsonArchivoCuentaRepository(CuentaRepositoryPort):
    def __init__(self, ruta_archivo: str):
        self.ruta = ruta_archivo
        self._datos: Dict[str, dict] = {}
        self._cargar()

    def _cargar(self):
        try:
            with open(self.ruta, "r", encoding="utf-8") as f:
                self._datos = json.load(f)
        except (FileNotFoundError, json.JSONDecodeError):
            self._datos = {}

    def _guardar_disco(self):
        with open(self.ruta, "w", encoding="utf-8") as f:
            json.dump(self._datos, f, indent=2)

    def obtener_por_numero(self, numero: str) -> Optional[CuentaBancaria]:
        item = self._datos.get(numero)
        if not item: return None
        return CuentaBancaria(item["numero"], item["titular"], Decimal(str(item["saldo"])), Decimal(str(item["descubierto"])))

    def guardar(self, cuenta: CuentaBancaria) -> None:
        self._datos[cuenta.numero_cuenta] = {
            "numero": cuenta.numero_cuenta, "titular": cuenta.titular,
            "saldo": float(cuenta.saldo), "descubierto": float(cuenta.limite_descubierto)
        }
        self._guardar_disco()

print("Dos adaptadores listos. Ambos cumplen el contrato 'CuentaRepositoryPort'.")
````

### 1. ¿Cómo estructurar los paquetes en el disco?
Simon Brown y Uncle Bob analizan 4 formas de organizar un proyecto:
1. **Package by Layer (Por Capa)**: `controllers/`, `services/`, `models/`, `repositories/`.
   - *Problema:* No grita el dominio y rompe el encapsulamiento fácilmente.
2. **Package by Feature (Por Funcionalidad)**: `cuentas/`, `prestamos/`, `facturas/`.
   - *Mejora:* Grita el dominio de negocio, pero aún puede mezclar la base de datos con la lógica.
3. **Package by Component (Recomendado)**:
   Agrupar cada componente funcional con límites estrictos de acceso:
   ```
   src/
   ├── core/                           # (Círculo Interior)
   │   ├── domain/                     # Entidades puras (CuentaBancaria, Prestamo)
   │   └── use_cases/                  # Casos de uso e Interfaces de Puertos
   │       ├── transferir/
   │       └── abrir_cuenta/
   ├── adapters/                       # (Círculo Intermedio)
   │   ├── repositories/               # PostgresRepo, MongoRepo, InMemRepo
   │   ├── presenters/                 # CliPresenter, JsonPresenter
   │   └── controllers/                # HttpControllers, CliControllers
   └── main.py                         # (Composition Root: cablea todo)
   ```
